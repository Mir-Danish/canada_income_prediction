```python
import pandas as pd
import numpy as np
from sklearn import linear_model
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('revenue.csv')
df.head()
```

```python
%matplotlib inline
plt.xlabel('marketing_spend')
plt.ylabel('monthly_revenue')
#plt.show()
plt.scatter(df.marketing_spend,df.monthly_revenue)
```

```python
regr = linear_model.LinearRegression()
regr.fit(df[['marketing_spend']],df.monthly_revenue)
```

```python
Y = pd.DataFrame({"marketing_spend": [60]})
regr.predict(Y)
```

```python
%matplotlib inline
plt.xlabel('marketing_spend', fontsize=20)
plt.ylabel('monthly_revenue', fontsize=20)
plt.scatter(df.marketing_spend,df.monthly_revenue, color='black')
plt.plot(df.marketing_spend,regr.predict(df[['marketing_spend']]),color='blue')
```
